```python
#Semicolons are optional in Python. You can write multiple statements on one line by separating them with ; but this is rarely used because it makes it hard to read:
#Text in Python must be inside quotes. 
#You can use either " double quotes or ' single quotes:


print("Hello"); print("How are you?"); print("Bye bye!")
```

```python
#We can print numbers also. 
#However, unlike text, we don't put numbers inside double quotes:
print(3)
print(358)
print(50000)
```

```python
#Mixed Numbers and Text printing
print("I am", 35, "years old.")
```

```python
#Python has no command for declaring a variable.
#A variable is created the moment you first assign a value to it.
#Variables do not need to be declared with any particular type, and can even change type after they have been set.

x = 4       # x is of type int
x = "Sally" # x is now of type str
print(x)
```

```python
#If you want to specify the data type of a variable, this can be done with casting.

x = str(3)    # x will be '3'
y = int(3)    # y will be 3
z = float(3)  # z will be 3.0
```

```python
#You can get the data type of a variable with the type() function.
x = 5
y = "John"
print(type(x))
print(type(y))
```
