#Python - Global Variables

```python
#Variables that are created outside of a function are known as global variables.

#Global variables can be used by everyone, both inside of functions and outside.

#Create a variable outside of a function, and use it inside the function

x = "awesome"

def myfunc():
  print("Python is " + x)

myfunc()
```

```python
#If you create a variable with the same name inside a function, 
#this variable will be local, and can only be used inside the function. 
#The global variable with the same name will remain as it was, global and 
#with the original value.

x = "awesome"
def myfunc():
     x = "fantastic"
     print("Python is " + x)
myfunc()
print("Python is " + x)  
```

```python

#Normally, when you create a variable inside a function, that variable is local, 
#and can only be used inside that function.

#To create a global variable inside a function, you can use the global keyword.

x = "good"
def myfunc():
    global x
    x = "nice"
    myfunc()    
print("Python is " + x)    
```

```python
x = 'awesome'
def myfunc():
  global x
  x = 'fantastic'
  myfunc()
print('Python is ' + x)

#However, if the indendation is placed at wrong place, it might give different result.
#It maynot give error but the output will be different. 
# This is because the function is not called before printing the value of x.
#This will

y = 'awesome'
def myfunc():
  global y
  y = 'fantastic'
myfunc()
print('Python is ' + y)
```

```python
#In Python, the global keyword allows you to modify a variable outside of the 
#current function's scope. Let's break down what is actually happening in both scenarios.

#Case 1: The Recursive Loop (Potential Crash)
#In first example, the indentation of myfunc() inside the definition creates a recursive loop.


x = 'awesome'
def myfunc():
  global x
  x = 'fantastic'
  myfunc() # This is INSIDE the function
print('Python is ' + y)
#What happens: When you call myfunc(), it sets x to 'fantastic', but then it calls itself immediately. It will keep calling itself until Python hits a RecursionError (maximum recursion depth exceeded).

#The Result: If you tried to print x after this, the program might have crashed before reaching the print statement. If it did manage to print, it would be "Python is fantastic".

#Case 2: The Standard Global Update
#In second example, the function is called once, and the variable is updated.


y = 'awesome'
def myfunc():
  global y
  y = 'fantastic'

myfunc() # This is OUTSIDE the function
print('Python is ' + y)
#What happens: You define the function, then you call it once. The global y line tells Python, "Don't create a new local y; use the one from the main script." It changes 'awesome' to 'fantastic'.

#The Result: The output will be "Python is fantastic".

#Why you might see "Awesome" or "Good"
#We see "Python is awesome" in the first example, it’s likely because the function 
# was never actually executed. In Python, simply defining a function 
# (def myfunc():) doesn't run the code inside it. You have to call it.

#If we don't call myfunc(), x remains 'awesome'.

#If you do call myfunc() (and it has global x), x becomes 'fantastic'.
```
