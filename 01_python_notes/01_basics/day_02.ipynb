# Python Data Types

## Built-in Data Types
In programming, a data type defines the kind of values a variable can hold and how they are used.

Variables can store data of different types, and different types support different operations.
Python has the following data types built-in by default, in these categories:

1. Text Type: str — A sequence of Unicode characters used to represent textual data.
2. Numeric Types:
   - int — Represents whole numbers of arbitrary size.
   - float — Represents real numbers using double-precision floating point.
   - complex — Represents complex numbers with real and imaginary parts.
3. Sequence Types:
   - list — An ordered, mutable collection of items.
   - tuple — An ordered, immutable collection of items.
   - range — Represents an immutable sequence of integers commonly used for looping.
4. Mapping Type: dict — An unordered, mutable collection of key-value pairs for fast lookup by key.
5. Set Types:
   - set — An unordered collection of unique, mutable elements.
   - frozenset — An unordered collection of unique elements that is immutable.
6. Boolean Type: bool — Represents truth values, either True or False.
7. Binary Types:
   - bytes — An immutable sequence of bytes.
   - bytearray — A mutable sequence of bytes.
   - memoryview — A view that exposes the memory of another binary object without copying it.
8. None Type: NoneType — The type of the singleton value None used to represent the absence of a value.

```python
# One-line code examples for each built-in data type
s = "hello"                # str
i = 42                     # int
f = 3.14                   # float
c = 1+2j                   # complex
lst = [1, 2, 3]            # list
tup = (1, 2, 3)            # tuple
rng = range(3)             # range
d = {"a": 1}               # dict
st = {1, 2, 3}             # set
fs = frozenset({1, 2})     # frozenset
b = True                   # bool
by = b"bytes"             # bytes
ba = bytearray(b"abc")    # bytearray
mv = memoryview(b"abc")   # memoryview
n = None                   # NoneType
```

```python
# Print type and value for each example variable
for name in ['s','i','f','c','lst','tup','rng','d','st','fs','b','by','ba','mv','n']:
    var = eval(name)
    print(f"{name}: {var} ({type(var).__name__})")
```
